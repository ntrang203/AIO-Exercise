### Bài 1 Gradient Descent:

```python
import numpy as np
```

```python
def df_w(W):
    # Unpack w1 and w2 from W
    w1, w2 = W

    # Compute partial derivatives
    dw1 = 0.2 * w1  # ∂f/∂w1
    dw2 = 4 * w2  # ∂f/∂w2

    # Create gradient array
    dW = np.array([dw1, dw2])

    return dW
```

```python
def sgd(W, dW, lr):
    W = W - lr*dW
    return W
```

```python
def train_p1(optimizer, lr, epochs):
  # Initial point
    W = np.array([-5, -2], dtype=np.float32)
    # List of results
    results = [W.copy()]
    for epoch in range(epochs):
        # Compute the gradient at the current point
        dW = df_w(W)  # Assume `df_w` is defined to compute gradients

        # Update W using the optimizer
        W = optimizer(W, dW, lr)

        # Append the updated W to results
        results.append(W.copy())

    return results
```

```python
train_p1(sgd, lr=0.4, epochs=30)
```

### Bài 2 Gradient Descent + Momentum:

```python
def df_w(W):
    # Unpack w1 and w2 from W
    w1, w2 = W

    # Compute partial derivatives
    dw1 = 0.2 * w1  # ∂f/∂w1
    dw2 = 4 * w2  # ∂f/∂w2

    # Create gradient array
    dW = np.array([dw1, dw2])

    return dW
```

```python
def sgd_momentum(W, dW, lr, V, beta):
    V = beta*V + (1-beta)*dW
    W = W - lr*V

    return W, V
```

```python
def train_p1(optimizer, lr, epochs, beta=0.9): # Added beta parameter with default value
    # initial
    W = np.array([-5, -2], dtype=np.float32)
    V = np.array([0, 0], dtype=np.float32)
    results = [W.copy()]

    for epoch in range(epochs):
        # Compute the gradient at the current point
        dW = df_w(W)  # Gradient calculation (assumes `df_w` is defined)

        # Update W and V using the optimizer, now passing beta
        W, V = optimizer(W, dW, lr, V, beta)

        # Store the updated W
        results.append(W.copy())

    return results
```

```python
train_p1(sgd_momentum, lr=0.6, epochs=30)
```

### Bài 3 RMSProp:

```python
def df_w(W):
    # Unpack w1 and w2 from W
    w1, w2 = W

    # Compute partial derivatives
    dw1 = 0.2 * w1  # ∂f/∂w1
    dw2 = 4 * w2  # ∂f/∂w2

    # Create gradient array
    dW = np.array([dw1, dw2])

    return dW
```

```python
def RMSProp(W, dW, lr, S, gamma):
    epsilon = 1e-6

    # Update the exponentially weighted average of squared gradients
    S = gamma * S + (1 - gamma) * (dW ** 2)

    # Update the weights using RMSProp formula
    W = W - lr * dW / (np.sqrt(S) + epsilon)

    return W, S
```

```python
def train_p1(optimizer, lr, epochs):
    # Initial weights and moving averages
    W = np.array([-5, -2], dtype=np.float32)
    S = np.array([0, 0], dtype=np.float32)  # Initialize moving averages to zero

    # Store results
    results = [W.copy()]

    # Optimization loop
    for epoch in range(epochs):
        # Compute gradients
        dW = df_w(W)  # Assuming `df_w` is defined to calculate gradients

        # Update weights and moving averages using the optimizer
        W, S = optimizer(W, dW, lr, S, gamma=0.9)  # Pass decay rate gamma (e.g., 0.9)

        # Store updated weights
        results.append(W.copy())

    return results
```

```python
train_p1(RMSProp, lr=0.3, epochs=30)
```

### Bài 4 Adam:

```python
def df_w(W):
    # Unpack w1 and w2 from W
    w1, w2 = W

    # Compute partial derivatives
    dw1 = 0.2 * w1  # ∂f/∂w1
    dw2 = 4 * w2  # ∂f/∂w2

    # Create gradient array
    dW = np.array([dw1, dw2])

    return dW
```

```python
def Adam(W, dW, lr, V, S, beta1, beta2, t):
    epsilon = 1e-6

    # Update exponentially weighted averages
    V = beta1 * V + (1 - beta1) * dW
    S = beta2 * S + (1 - beta2) * (dW ** 2)

    # Compute bias-corrected first and second moments
    V_corrected = V / (1 - beta1 ** t)
    S_corrected = S / (1 - beta2 ** t)

    # Update weights
    W = W - lr * V_corrected / (np.sqrt(S_corrected) + epsilon)

    return W, V, S
```

```python
def train_p1(optimizer, lr, epochs):
    # Initial values
    W = np.array([-5, -2], dtype=np.float32)
    V = np.array([0, 0], dtype=np.float32)  # First moment vector
    S = np.array([0, 0], dtype=np.float32)  # Second moment vector
    results = [W.copy()]

    # Optimization loop
    for t in range(1, epochs + 1):  # t starts at 1
        # Calculate gradient
        dW = df_w(W)  # Assuming `df_w` is defined elsewhere

        # Update weights using the optimizer
        W, V, S = optimizer(W, dW, lr, V, S, beta1=0.9, beta2=0.999, t=t)

        # Store updated weights
        results.append(W.copy())

    return results
```

```python
train_p1(Adam, lr=0.2, epochs=30)
```

```python
import torch
from torch import nn
from adopt import ADOPT
from torch.utils.data import Dataset, DataLoader
import torchvision
from torchvision.datasets import FashionMNIST
import torchvision.transforms as transforms
import numpy as np
device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
torch.manual_seed(42)
```

```python
batch_size = 512
num_epochs = 300
lr = 0.001
```

```python
train_dataset = FashionMNIST(root='./data', train=True, download=True, transform=transforms.ToTensor())
train_loader = DataLoader(train_dataset, batch_size, shuffle=True)
test_dataset = FashionMNIST(root='./data', train=False, download=True, transform=transforms.ToTensor())
test_loader = DataLoader(test_dataset, batch_size)
```

### Build và Compile Model

```python
class MLP(nn.Module):
    def __init__(self, input_dims, hidden_dims, output_dims):
        super(MLP, self).__init__()
        self.layer1 = nn.Linear(input_dims, hidden_dims)
        self.layer2 = nn.Linear(hidden_dims, hidden_dims)
        self.layer3 = nn.Linear(hidden_dims, hidden_dims)
        self.layer4 = nn.Linear(hidden_dims, hidden_dims)
        self.layer5 = nn.Linear(hidden_dims, hidden_dims)
        self.output = nn.Linear(hidden_dims, output_dims)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = nn.Flatten()(x)
        x = self.layer1(x)
        x = self.sigmoid(x)
        x = self.layer2(x)
        x = self.sigmoid(x)
        x = self.layer3(x)
        x = self.sigmoid(x)
        x = self.layer4(x)
        x = self.sigmoid(x)
        x = self.layer5(x)
        x = self.sigmoid(x)
        out = self.output(x)
        return out
```

```python
import torch.optim as optim

model = MLP(input_dims=784, hidden_dims=128, output_dims=10).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=lr)
```

```python
train_losses = []
train_acc = []
val_losses = []
val_acc = []
for epoch in range(num_epochs):
    model.train()
    t_loss = 0
    t_acc = 0
    cnt = 0
    for X, y in train_loader:
        X, y = X.to(device), y.to(device)
        optimizer.zero_grad()
        outputs = model(X)
        loss = criterion(outputs, y)
        loss.backward()
        optimizer.step()
        t_loss += loss.item()
        t_acc += (torch.argmax(outputs, 1) == y).sum().item()
        cnt += len(y)
    t_loss /= len(train_loader)
    train_losses.append(t_loss)
    t_acc /= cnt
    train_acc.append(t_acc)

    model.eval()
    v_loss = 0
    v_acc = 0
    cnt = 0
    with torch.no_grad():
        for X, y in test_loader:
            X, y = X.to(device), y.to(device)
            outputs = model(X)
            loss = criterion(outputs, y)
            v_loss += loss.item()
            v_acc += (torch.argmax(outputs, 1)==y).sum().item()
            cnt += len(y)
    v_loss /= len(test_loader)
    val_losses.append(v_loss)
    v_acc /= cnt
    val_acc.append(v_acc)
    print(f"Epoch {epoch+1}/{num_epochs}, Train_Loss: {t_loss:.4f}, Train_Acc: {t_acc:.4f}, Validation Loss: {v_loss:.4f}, Val_Acc: {v_acc:.4f}")
```
